```python
import random
import sympy as sp

# Funkcija generuoti kvadratinę lygtį
def generate_quadratic_equation():
    a = random.randint(1, 10)
    b = random.randint(-10, 10)
    c = random.randint(-10, 10)

    # Sukuriame kvadratinę lygtį
    equation = f"{a}x^2 + {b}x + {c} = 0"

    # Sprendžiame kvadratinę lygtį
    x = sp.symbols('x')
    quadratic = a*x**2 + b*x + c
    solutions = sp.solve(quadratic, x)

    return equation, solutions

# Funkcija generuoti trikampio plotą
def generate_triangle_area():
    base = random.randint(5, 15)
    height = random.randint(5, 15)
    area = 0.5 * base * height
    return f"Trikampio, kurio pagrindas {base} cm ir aukštinė {height} cm, koks būtų šio trikampio plotas?", area

# Funkcija generuoti geometrinę progresiją
def generate_geometric_sequence():
    a1 = random.randint(1, 10)  # Pirmasis narys
    r = random.randint(2, 5)    # Progresijos šuolis
    n = random.randint(3, 7)    # Narių skaičius

    # Generuojame seką
    sequence = [a1 * (r ** i) for i in range(n)]
    return f"Geometrinė progresija su pirmuoju nariu {a1} ir skirtumu {r}:", sequence

# Pagrindinė funkcija, kuri sukuria kelias užduotis
def generate_math_problems():
    problems = []

    # Generuojame kvadratinę lygtį
    eq, solutions = generate_quadratic_equation()
    problems.append(f"Užduotis: Išspręskite šią kvadratinę lygtį: {eq}\nAtsakymai: {solutions}")

    # Generuojame trikampio plotą
    area_problem, area_solution = generate_triangle_area()
    problems.append(f"Užduotis: {area_problem}\nAtsakymas: {area_solution} cm²")

    # Generuojame geometrinę progresiją
    sequence_problem, sequence = generate_geometric_sequence()
    problems.append(f"Užduotis: {sequence_problem} {sequence}")

    return problems

# Spausdiname užduotis
generated_problems = generate_math_problems()

# Išvedame sukurtas užduotis
for problem in generated_problems:
    print(problem)
    print("-" * 50)

```
